```python
num_column=[['GDP (nominal, 2023)','GDP (abbrev.)','GDP Growth','Population 2023','GDP per capita','Share of World GDP']]
```

```python
df1['GDP Growth'] = df1['GDP Growth'].str.replace('%', '').str.replace('−', '-').astype(float) / 100
```

```python
df1['GDP per capita'] = df1['GDP per capita'].astype(str).str.replace('$', '').str.replace(',', '')
```

```python
df1['Share of World GDP'] = pd.to_numeric(df1['Share of World GDP'].str.replace('%',' '),errors='coerce')
```

```python
df1.drop(columns='GDP (abbrev.)', inplace=True)
```
